```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
# Import random module
import random

# Set seed for reproducible results
np.random.seed(42)
random.seed(42)

# Department-wise employee requirements
departments = {
    "Engineering": 50,
    "Data & AI": 15,
    "Sales & Marketing": 20,
    "Customer Success": 15,
    "HR & Administration": 10,
    "Finance & Operations": 10
}

# Job roles for each department
roles = {
    "Engineering": ["Software Engineer", "Frontend Developer", "Backend Developer"],
    "Data & AI": ["Data Analyst", "AI/ML Engineer", "Data Engineer"],
    "Sales & Marketing": ["Sales Executive", "Marketing Executive"],
    "Customer Success": ["Customer Success Executive", "Support Specialist"],
    "HR & Administration": ["HR Executive", "Recruiter", "Administrator"],
    "Finance & Operations": ["Financial Analyst", "Operations Executive"]
}

# Create employee records
employee_data = []

employee_id = 1001

for department, count in departments.items():

    for i in range(count):

        employee_data.append({
            "Employee_ID": employee_id,
            "Department": department,
            "Job_Role": random.choice(roles[department]),
            "Experience_Years": random.randint(0, 12),
            "Performance_Score": round(np.random.uniform(2.5, 5.0), 1),
            "Attrition_Risk": random.choice(["Low", "Medium", "High"])
        })

        employee_id += 1

# Convert data into DataFrame
employees_df = pd.DataFrame(employee_data)

# Display first 10 employees
employees_df.head(10)
```

```python
# Workforce Gap Analysis

# Future workforce requirements
future_requirements = {
    "Engineering": 70,
    "Data & AI": 25,
    "Sales & Marketing": 27,
    "Customer Success": 20,
    "HR & Administration": 13,
    "Finance & Operations": 10
}

# Count current employees department-wise
current_workforce = employees_df.groupby("Department").size()

# Create workforce analysis table
workforce_analysis = pd.DataFrame({
    "Current_Employees": current_workforce,
    "Future_Requirement": pd.Series(future_requirements)
})

# Calculate workforce gap
workforce_analysis["Hiring_Gap"] = (
    workforce_analysis["Future_Requirement"] -
    workforce_analysis["Current_Employees"]
)

# Display the result
workforce_analysis
```

```python
# Calculate total employees and total hiring requirement

total_current = workforce_analysis["Current_Employees"].sum()

total_future = workforce_analysis["Future_Requirement"].sum()

total_hiring_needed = workforce_analysis["Hiring_Gap"].sum()

print("WORKFORCE PLANNING SUMMARY")
print("-" * 35)

print("Current Workforce:", total_current)
print("Future Workforce Requirement:", total_future)
print("Total New Hires Required:", total_hiring_needed)
```

```python
# Workforce Hiring Gap Visualization

plt.figure(figsize=(10, 6))

plt.bar(
    workforce_analysis.index,
    workforce_analysis["Hiring_Gap"]
)

plt.title("Department-wise Hiring Gap Analysis")
plt.xlabel("Department")
plt.ylabel("Employees Required")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
# Current vs Future Workforce Comparison

plt.figure(figsize=(10, 6))

x = np.arange(len(workforce_analysis))

width = 0.35

plt.bar(
    x - width/2,
    workforce_analysis["Current_Employees"],
    width,
    label="Current Workforce"
)

plt.bar(
    x + width/2,
    workforce_analysis["Future_Requirement"],
    width,
    label="Future Requirement"
)

plt.xticks(
    x,
    workforce_analysis.index,
    rotation=30
)

plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.title("Current Workforce vs Future Requirement")

plt.legend()
```

```python
# Candidate Recruitment Dataset

np.random.seed(42)

candidate_data = {
    "Candidate_ID": range(1, 31),

    "Role_Applied": np.random.choice([
        "Software Engineer",
        "AI/ML Engineer",
        "Data Analyst",
        "Cloud/DevOps Engineer"
    ], 30),

    "Experience_Years": np.random.randint(0, 8, 30),

    "Technical_Skills": np.random.randint(40, 101, 30),

    "Problem_Solving": np.random.randint(40, 101, 30),

    "Communication": np.random.randint(40, 101, 30),

    "Teamwork": np.random.randint(40, 101, 30),

    "Adaptability": np.random.randint(40, 101, 30),

    "Values_Motivation": np.random.randint(40, 101, 30)
}

# Convert into DataFrame
candidates_df = pd.DataFrame(candidate_data)

# Display first 10 candidates
candidates_df.head(10)
```

```python
# Calculate Weighted Candidate Score

candidates_df["Final_Score"] = (
    candidates_df["Technical_Skills"] * 0.35 +
    candidates_df["Problem_Solving"] * 0.20 +
    candidates_df["Communication"] * 0.15 +
    candidates_df["Teamwork"] * 0.10 +
    candidates_df["Adaptability"] * 0.10 +
    candidates_df["Values_Motivation"] * 0.10
)

# Round scores to 2 decimal places
candidates_df["Final_Score"] = candidates_df["Final_Score"].round(2)

# Display results
candidates_df[[
    "Candidate_ID",
    "Role_Applied",
    "Experience_Years",
    "Final_Score"
]].head(10)
```

```python
# Function to generate hiring recommendation

def get_recommendation(score):

    if score >= 75:
        return "Recommended"

    elif score >= 60:
        return "Consider"

    else:
        return "Not Selected"


# Apply recommendation to every candidate
candidates_df["Recommendation"] = candidates_df["Final_Score"].apply(
    get_recommendation
)

# Display candidate results
candidates_df[[
    "Candidate_ID",
    "Role_Applied",
    "Final_Score",
    "Recommendation"
]].head(15)
```

```python
# Recruitment Selection Summary

selection_summary = candidates_df["Recommendation"].value_counts()

print("RECRUITMENT SELECTION SUMMARY")
print("-" * 35)

print(selection_summary)
```

```python
# Candidate Recommendation Distribution

recommendation_counts = candidates_df["Recommendation"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    recommendation_counts.index,
    recommendation_counts.values
)

plt.title("Candidate Recruitment Recommendation Summary")
plt.xlabel("Recommendation")
plt.ylabel("Number of Candidates")

plt.tight_layout()

plt.show()
```

```python
# Role-wise Average Candidate Score

role_performance = (
    candidates_df
    .groupby("Role_Applied")["Final_Score"]
    .mean()
    .sort_values(ascending=False)
)

print("ROLE-WISE AVERAGE CANDIDATE SCORE")
print("-" * 40)

print(role_performance.round(2))
```

```python
# Role-wise Candidate Performance Chart

plt.figure(figsize=(10, 6))

plt.bar(
    role_performance.index,
    role_performance.values
)

plt.title("Average Candidate Score by Role")
plt.xlabel("Role")
plt.ylabel("Average Final Score")

plt.xticks(rotation=25)

plt.tight_layout()

plt.show()
```

```python
# Top 10 Candidates

top_candidates = (
    candidates_df
    .sort_values(by="Final_Score", ascending=False)
    .head(10)
)

print("TOP 10 CANDIDATES")

top_candidates[[
    "Candidate_ID",
    "Role_Applied",
    "Experience_Years",
    "Final_Score",
    "Recommendation"
]]
```

```python
# Recruitment KPI Data

kpi_data = {
    "KPI": [
        "Time to Hire",
        "Qualified Applicant Rate",
        "Offer Acceptance Rate",
        "90-Day Retention",
        "Quality of Hire",
        "Cost per Hire"
    ],

    "Current_Value": [
        38,
        45,
        80,
        92,
        4.2,
        15000
    ],

    "Target_Value": [
        30,
        60,
        85,
        90,
        4.0,
        18000
    ]
}

# Create DataFrame
kpi_df = pd.DataFrame(kpi_data)

# Display KPI Dashboard Data
kpi_df
```

```python
# Function to calculate KPI status

def get_kpi_status(row):

    kpi = row["KPI"]
    current = row["Current_Value"]
    target = row["Target_Value"]

    # Lower value is better
    if kpi in ["Time to Hire", "Cost per Hire"]:

        if current <= target:
            return "Achieved"
        else:
            return "Needs Improvement"

    # Higher value is better
    else:

        if current >= target:
            return "Achieved"
        elif current >= target * 0.90:
            return "Close to Target"
        else:
            return "Needs Improvement"


# Apply function
kpi_df["Status"] = kpi_df.apply(get_kpi_status, axis=1)

# Display results
kpi_df
```

```python
# Recruitment KPI Performance Chart

plt.figure(figsize=(10, 6))

plt.bar(
    kpi_df["KPI"],
    kpi_df["Current_Value"]
)

plt.title("Recruitment KPI Current Performance")
plt.xlabel("KPI")
plt.ylabel("Current Value")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
# KPI Status Summary

status_summary = kpi_df["Status"].value_counts()

print("RECRUITMENT KPI STATUS SUMMARY")
print("-" * 35)

print(status_summary)
```

```python
# Quarterly Hiring Forecast Data

forecast_data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Opening_Workforce": [120, 132, 143, 153],
    "Planned_Hires": [14, 13, 11, 12],
    "Planned_Exits": [2, 2, 1, 0],
    "Closing_Workforce": [132, 143, 153, 165]
}

forecast_df = pd.DataFrame(forecast_data)

forecast_df
```

```python
# Verify Workforce Forecast Calculations

forecast_df["Calculated_Closing"] = (
    forecast_df["Opening_Workforce"]
    + forecast_df["Planned_Hires"]
    - forecast_df["Planned_Exits"]
)

forecast_df["Forecast_Status"] = np.where(
    forecast_df["Calculated_Closing"] == forecast_df["Closing_Workforce"],
    "Verified",
    "Check Calculation"
)

forecast_df
```

```python
# Quarterly Workforce Growth

plt.figure(figsize=(9, 5))

plt.plot(
    forecast_df["Quarter"],
    forecast_df["Closing_Workforce"],
    marker="o"
)

plt.title("Quarterly Workforce Growth Forecast")
plt.xlabel("Quarter")
plt.ylabel("Closing Workforce")

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
# Planned Hires vs Planned Exits

x = np.arange(len(forecast_df))

width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(
    x - width/2,
    forecast_df["Planned_Hires"],
    width,
    label="Planned Hires"
)

plt.bar(
    x + width/2,
    forecast_df["Planned_Exits"],
    width,
    label="Planned Exits"
)

plt.xticks(x, forecast_df["Quarter"])

plt.title("Quarterly Planned Hires vs Exits")
plt.xlabel("Quarter")
plt.ylabel("Number of Employees")

plt.legend()

plt.tight_layout()

plt.show()
```

```python
# Smart Recruitment Priority System

def get_priority(hiring_gap):

    if hiring_gap >= 15:
        return "Very High"

    elif hiring_gap >= 8:
        return "High"

    elif hiring_gap >= 3:
        return "Medium"

    else:
        return "Low"


# Create priority column
workforce_analysis["Hiring_Priority"] = workforce_analysis[
    "Hiring_Gap"
].apply(get_priority)

# Sort departments by hiring gap
priority_analysis = workforce_analysis.sort_values(
    by="Hiring_Gap",
    ascending=False
)

print("SMART RECRUITMENT PRIORITY ANALYSIS")

priority_analysis
```

```python
# Calculate Recruitment Priority Score

total_hiring_required = workforce_analysis["Hiring_Gap"].sum()

priority_analysis["Priority_Score"] = (
    priority_analysis["Hiring_Gap"]
    / total_hiring_required
    * 100
).round(2)

priority_analysis
```

```python
# Recruitment Priority Visualization

plt.figure(figsize=(10, 6))

plt.bar(
    priority_analysis.index,
    priority_analysis["Priority_Score"]
)

plt.title("Smart Recruitment Priority by Department")
plt.xlabel("Department")
plt.ylabel("Priority Score (%)")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
# Identify Top Recruitment Priority

top_department = priority_analysis.index[0]

top_gap = priority_analysis.iloc[0]["Hiring_Gap"]

top_priority = priority_analysis.iloc[0]["Hiring_Priority"]

print("TOP RECRUITMENT PRIORITY")
print("-" * 35)

print("Department:", top_department)
print("Hiring Gap:", top_gap)
print("Priority Level:", top_priority)
```

```python
# FINAL PROJECT MANAGEMENT SUMMARY

current_workforce_total = workforce_analysis["Current_Employees"].sum()

future_workforce_total = workforce_analysis["Future_Requirement"].sum()

total_hiring_required = workforce_analysis["Hiring_Gap"].sum()

recommended_candidates = (
    candidates_df["Recommendation"] == "Recommended"
).sum()

consider_candidates = (
    candidates_df["Recommendation"] == "Consider"
).sum()

not_selected_candidates = (
    candidates_df["Recommendation"] == "Not Selected"
).sum()


print("=" * 50)
print("RECRUITMENT STRATEGY & WORKFORCE PLANNING")
print("FINAL MANAGEMENT SUMMARY")
print("=" * 50)

print("\nWORKFORCE SUMMARY")
print("Current Workforce:", current_workforce_total)
print("Future Workforce Requirement:", future_workforce_total)
print("Total Hiring Required:", total_hiring_required)

print("\nRECRUITMENT PRIORITY")
print("Top Priority Department:", top_department)
print("Hiring Gap:", top_gap)
print("Priority Level:", top_priority)

print("\nCANDIDATE SELECTION SUMMARY")
print("Recommended:", recommended_candidates)
print("Consider:", consider_candidates)
print("Not Selected:", not_selected_candidates)

print("=" * 50)
```
